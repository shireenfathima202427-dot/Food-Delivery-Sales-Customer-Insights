```python
df = pd.read_csv("data/order_history_kaggle_data.csv")

print(df['Order Placed At'].head(10).tolist())
```

```python
df['Order Placed At'] = pd.to_datetime(
    df['Order Placed At'],
    format='mixed'
)
```

```python
df['Hour'] = df['Order Placed At'].dt.hour
df['Day'] = df['Order Placed At'].dt.day_name()
df['Month'] = df['Order Placed At'].dt.month_name()

df.head()
```

```python
df.isnull().sum()
```

```python
# Text columns
df['Instructions'] = df['Instructions'].fillna('No Instructions')
df['Review'] = df['Review'].fillna('No Review')
df['Customer complaint tag'] = df['Customer complaint tag'].fillna('No Complaint')

# Discount column
df['Discount construct'] = df['Discount construct'].fillna('No Discount')

# Rating
df['Rating'] = df['Rating'].fillna(df['Rating'].mean())

# Delivery metrics
df['KPT duration (minutes)'] = df['KPT duration (minutes)'].fillna(
    df['KPT duration (minutes)'].median()
)

df['Rider wait time (minutes)'] = df['Rider wait time (minutes)'].fillna(
    df['Rider wait time (minutes)'].median()
)
```

```python
df['Cancellation / Rejection reason'] = (
    df['Cancellation / Rejection reason']
    .fillna('Not Cancelled')
)

df['Restaurant compensation (Cancellation)'] = (
    df['Restaurant compensation (Cancellation)']
    .fillna(0)
)

df['Restaurant penalty (Rejection)'] = (
    df['Restaurant penalty (Rejection)']
    .fillna(0)
)
```

```python
df.isnull().sum()
```

```python
duplicates = df.duplicated().sum()
print("Duplicate Rows:", duplicates)

df.drop_duplicates(inplace=True)
```

```python
print("Total Orders:", len(df))
print("Total Revenue:", df['Total'].sum())
print("Top City:", df['City'].mode()[0])
print("Top Restaurant:", df['Restaurant name'].mode()[0])
print("Average Rating:", round(df['Rating'].mean(),2))
print("Most Common Status:", df['Order Status'].mode()[0])
```

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("data/order_history_kaggle_data.csv")

city_orders = df['City'].value_counts().head(10)

plt.figure(figsize=(8, 8))

plt.pie(
    city_orders.values,
    labels=city_orders.index,
    autopct='%1.1f%%',
    startangle=140
)

# Create hole in center (DONUT)
centre_circle = plt.Circle((0, 0), 0.60, fc='white')
fig = plt.gcf()
fig.gca().add_artist(centre_circle)

plt.title("Top 10 Cities - Doughnut Chart", fontsize=14, fontweight='bold')
plt.show()
```

```python
import matplotlib.pyplot as plt

status_counts = df['Order Status'].value_counts()

plt.figure(figsize=(8, 5))

plt.bar(status_counts.index, status_counts.values)

plt.title("Order Status Distribution", fontsize=14, fontweight='bold')
plt.xlabel("Order Status")
plt.ylabel("Count")

plt.xticks(rotation=30)

plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt

restaurant_orders = df['Restaurant name'].value_counts().head(5)

plt.figure(figsize=(8, 8))

plt.pie(
    restaurant_orders,
    labels=None,        # remove text from slices
    autopct=None,       # remove numbers completely
    startangle=90
)

plt.title("Top 5 Restaurants by Orders", fontsize=14, fontweight='bold')

plt.legend(
    restaurant_orders.index,
    title="Restaurants",
    loc="center left",
    bbox_to_anchor=(1, 0.5)
)

plt.axis('equal')
plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt

restaurant_revenue = (
    df.groupby('Restaurant name')['Total']
      .sum()
      .sort_values(ascending=False)
      .head(10)
)

plt.figure(figsize=(12, 5))

colors = plt.cm.tab10(range(len(restaurant_revenue)))

plt.bar(
    restaurant_revenue.index,
    restaurant_revenue.values,
    color=colors
)

plt.title("Top 10 Restaurants by Revenue", fontsize=14, fontweight='bold')
plt.xlabel("Restaurant")
plt.ylabel("Revenue")

plt.xticks(rotation=45)
plt.tight_layout()

plt.show()
```

```python
plt.figure(figsize=(8,5))

plt.hist(df['Rating'], bins=10)

plt.title("Customer Rating Distribution")
plt.xlabel("Rating")
plt.ylabel("Frequency")

plt.show()
```

```python
restaurant_data = (
    df.groupby('Restaurant name')
      .agg({'Rating':'mean','Total':'sum'})
)

plt.figure(figsize=(8,6))

plt.scatter(
    restaurant_data['Rating'],
    restaurant_data['Total']
)

plt.title("Rating vs Revenue")
plt.xlabel("Average Rating")
plt.ylabel("Revenue")

plt.show()
```

```python
distance_counts = df['Distance'].value_counts().head(10)

plt.figure(figsize=(10,5))

plt.bar(distance_counts.index, distance_counts.values)

plt.title("Most Common Delivery Distances")
plt.xlabel("Distance")
plt.ylabel("Number of Orders")

plt.show()
```

```python
top_customers = df['Customer ID'].value_counts().head(10)

short_ids = [f'C{i+1}' for i in range(len(top_customers))]

colors = [
    'red', 'orange', 'gold', 'green', 'cyan',
    'blue', 'purple', 'pink', 'brown', 'gray'
]

plt.figure(figsize=(10,5))

bars = plt.bar(
    short_ids,
    top_customers.values,
    color=colors
)

# Add values on top of bars
for bar in bars:
    plt.text(
        bar.get_x() + bar.get_width()/2,
        bar.get_height(),
        int(bar.get_height()),
        ha='center',
        va='bottom',
        fontweight='bold'
    )

plt.title("Top 10 Customers by Number of Orders", fontsize=14, fontweight='bold')
plt.xlabel("Customer")
plt.ylabel("Orders")

plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

numeric_cols = [
    'Bill subtotal',
    'Total',
    'Rating',
    'KPT duration (minutes)',
    'Rider wait time (minutes)'
]

# Keep only valid numeric data (prevents errors)
corr_matrix = df[numeric_cols].apply(pd.to_numeric, errors='coerce').corr()

plt.figure(figsize=(10,6))

sns.heatmap(
    corr_matrix,
    annot=True,
    cmap='coolwarm',
    fmt=".2f",
    linewidths=0.5,
    cbar=True
)

plt.title("Correlation Heatmap", fontsize=14, fontweight='bold')
plt.xticks(rotation=30)
plt.yticks(rotation=0)

plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt

fig, ax = plt.subplots(2, 2, figsize=(16, 10))

# 1. Peak Order Hours
hourly_orders.plot(
    ax=ax[0,0],
    marker='o'
)
ax[0,0].set_title("Peak Order Hours")
ax[0,0].set_xlabel("Hour")
ax[0,0].set_ylabel("Orders")
ax[0,0].grid(True)

# 2. Top Cities (Donut Chart)
city_orders = df['City'].value_counts().head(10)

ax[0,1].pie(
    city_orders.values,
    labels=city_orders.index,
    autopct='%1.1f%%',
    startangle=140
)

centre_circle = plt.Circle((0,0), 0.60, fc='white')
ax[0,1].add_artist(centre_circle)

ax[0,1].set_title("Top Cities")

# 3. Order Status Distribution
status_counts = df['Order Status'].value_counts()

ax[1,0].bar(
    status_counts.index,
    status_counts.values
)

ax[1,0].set_title("Order Status Distribution")
ax[1,0].tick_params(axis='x', rotation=20)

# 4. Revenue by Restaurant
restaurant_revenue = (
    df.groupby('Restaurant name')['Total']
      .sum()
      .sort_values(ascending=False)
      .head(10)
)

colors = plt.cm.tab10(range(len(restaurant_revenue)))

ax[1,1].bar(
    restaurant_revenue.index,
    restaurant_revenue.values,
    color=colors
)

ax[1,1].set_title("Top Restaurants by Revenue")
ax[1,1].tick_params(axis='x', rotation=45)

plt.suptitle(
    "Food Delivery Sales & Customer Insights Dashboard",
    fontsize=18,
    fontweight='bold'
)

plt.tight_layout()
plt.show()
```
